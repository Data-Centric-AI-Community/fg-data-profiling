```python
import numpy as np
import pandas as pd
```

```python
array_1d = np.array([1, 2, 3, 4, 5])

array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:")
print(array_1d)
print("Shape:", array_1d.shape)

print("\n2D array:")
print(array_2d)
print("Shape:", array_2d.shape)

print("\n3D array:")
print(array_3d)
print("Shape:", array_3d.shape)

assert array_1d.shape == (5,)
assert array_2d.shape == (2, 3)
assert array_3d.shape == (2, 2, 2)

print("All array shapes verified successfully.")
```

```python
array = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

values = np.array([1, 2, 3])

result = array + values

print("Original 2D array:")
print(array)

print("\n1D array:")
print(values)

print("\nAfter broadcasting:")
print(result)
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([10, 20, 30, 40, 50])

addition = a + b
multiplication = a * b
square = a ** 2

print("Addition:", addition)
print("Multiplication:", multiplication)
print("Square:", square)
```

```python
matrix_a = np.array([
    [1, 2],
    [3, 4]
])

matrix_b = np.array([
    [5, 6],
    [7, 8]
])

element_wise = matrix_a * matrix_b
matrix_product = matrix_a @ matrix_b

print("Matrix A:")
print(matrix_a)

print("\nMatrix B:")
print(matrix_b)

print("\nElement-wise multiplication:")
print(element_wise)

print("\nMatrix multiplication:")
print(matrix_product)
```

```python
import numpy as np
import pandas as pd

file_path = "docs/getting-started/data_quality_alerts.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())

print("\nColumn names:")
print(df.columns.tolist())

# Select numeric columns
numeric_df = df.select_dtypes(include=np.number)

print("\nNumeric columns:")
print(numeric_df.columns.tolist())

# Mean using NumPy
mean_values = np.mean(numeric_df, axis=0)

# Standard deviation using NumPy
std_values = np.std(numeric_df, axis=0)

# Correlation using NumPy
correlation_matrix = np.corrcoef(numeric_df.T)

print("\nMean:")
print(mean_values)

print("\nStandard Deviation:")
print(std_values)

print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
import numpy as np
import pandas as pd

# Real Iris dataset
df_iris = pd.read_csv("iris.txt")

print("Dataset shape:", df_iris.shape)

print("\nFirst 5 rows:")
print(df_iris.head())

# Select numeric columns
numeric_data = df_iris.select_dtypes(include=np.number)

print("\nNumeric columns:")
print(numeric_data.columns.tolist())

# NumPy calculations
mean_values = np.mean(numeric_data, axis=0)
std_values = np.std(numeric_data, axis=0)
correlation_matrix = np.corrcoef(numeric_data.T)

print("\nMean:")
for column, value in zip(numeric_data.columns, mean_values):
    print(f"{column}: {value:.4f}")

print("\nStandard Deviation:")
for column, value in zip(numeric_data.columns, std_values):
    print(f"{column}: {value:.4f}")

print("\nCorrelation Matrix:")
print(pd.DataFrame(
    correlation_matrix,
    index=numeric_data.columns,
    columns=numeric_data.columns
))
```

## Viva Notes

### 1. What is broadcasting?
Broadcasting allows NumPy to perform operations between arrays of compatible shapes without explicitly copying or repeating the data. For example, adding a scalar to every element of an array.

### 2. Why are vectorized NumPy operations faster than Python loops?
NumPy operations are performed on complete arrays using optimized low-level code, which reduces the overhead of Python-level loops.

### 3. NumPy array vs Pandas Series
A NumPy array is a multidimensional numerical data structure used for efficient mathematical operations. A Pandas Series is a one-dimensional labeled data structure used for handling tabular data.
